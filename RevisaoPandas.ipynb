```python
array_inteiros = (10,20,30,40,50) # Criando uma Array
indices = ('a', 'b', 'c', 'd', 'e') # Criando índices

serie_dados = pd.Series(array_inteiros, index=indices) # Pegando o Array e o índice e transformando em uma Series
print(serie_dados)
```
